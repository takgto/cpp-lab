```python
from IPython.core.magic import register_cell_magic
import subprocess, sys

@register_cell_magic
def cpp(line, cell):
    open('tmp.cpp', 'w').write(cell)
    r = subprocess.run(['g++', '-O2', '-std=c++17', 'tmp.cpp', '-o', 'tmp'],
                       capture_output=True, text=True)
    if r.returncode:
        print(r.stderr, file=sys.stderr); return
    r = subprocess.run(['./tmp'] + line.split(), capture_output=True, text=True)
    print(r.stdout, end='')
    if r.stderr: print(r.stderr, file=sys.stderr, end='')
```
